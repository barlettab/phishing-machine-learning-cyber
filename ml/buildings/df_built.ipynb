```python
import pandas as pd

df = pd.read_csv("../datasets/raw/malicious_phish.csv")

df.head()
```

```python
df["type"].value_counts()
```

```python
df_benign = (
    df[df["type"] == "benign"]
    .copy()
)

df_benign["label"] = 0
df_benign["source"] = "kaggle_benign"
```

```python
df_benign.sample(
    20,
    random_state=42
)
```

```python
with open(
    "../datasets/raw/phishingGit.txt",
    "r",
    encoding="utf-8",
    errors="ignore"
) as f:

    urls = [
        linha.strip()
        for linha in f
        if linha.strip()
    ]

df_phishingdb = pd.DataFrame({
    "url": urls
})

df_phishingdb["label"] = 1
df_phishingdb["source"] = "phishingdb"
```

```python
df_phishingdb.shape
```

```python
print(len(df_phishingdb))
print(df_phishingdb["url"].nunique())
```

```python
df_phishingdb.sample(20, random_state=42)
```

```python
N = 57967

df_legit = (
    df_benign
    .sample(n=N, random_state=42)
)

df_phish = (
    df_phishingdb
    .sample(n=N, random_state=42)
)
```

```python
df_built = pd.concat(
    [df_legit, df_phish],
    ignore_index=True
)

df_built = df_built.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
df_built.head()
```

```python
df_built["label"].value_counts()
```

```python
df_built.drop(columns="type", inplace=True)

df_built.head()
```

```python
df_built.to_csv("../datasets/interim/df_built.csv", index=False)
```
