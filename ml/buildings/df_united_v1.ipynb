```python
# Carregando bibliotecas e o feature extractor v1
import pandas as pd
import sys
import os

sys.path.append(
    os.path.abspath("..")
)

from features.feature_extractor_v1 import extract_features
```

```python
# Carregando o df_built criado no notebook `df_built.ipynb`
df_built = pd.read_csv("../datasets/interim/df_built.csv")

# Aplicando a função extract_feature do feature_extractor_v1 no df_built
features_df = df_built["url"].apply(extract_features).apply(pd.Series)
```

```python
# Concatenar com o label
df_built_v1 = pd.concat(
    [
        features_df,
        df_built["label"]
    ],
    axis=1
)
```

```python
# Carregando o dataset_initial
df_initial = pd.read_csv("../datasets/raw/dataset_initial.csv")
```

```python
# Verificando a quantidade de linhas e colunas
df_initial.shape
```

```python
# Verificando as colunas existentes
print(df_initial.columns.tolist())

# Precisaremos retirar as colunas não correspondentes com as features que usaremos.
```

```python
# Utilizaremos somente as features associadas a URL 
url_features = [
    'NumDots',
    'SubdomainLevel',
    'PathLevel',
    'UrlLength',
    'NumDash',
    'NumDashInHostname',
    'AtSymbol',
    'TildeSymbol',
    'NumUnderscore',
    'NumPercent',
    'NumQueryComponents',
    'NumAmpersand',
    'NumHash',
    'NumNumericChars',
    'NoHttps',
    'IpAddress',
    'DomainInSubdomains',
    'DomainInPaths',
    'HostnameLength',
    'PathLength',
    'QueryLength',
    'DoubleSlashInPath',
]
```

```python
# Copiando o df_initial como df_initial_v1 para realizarmos as alterações nas colunas
df_initial_v1 = df_initial[
    url_features + ['CLASS_LABEL']
].copy()
```

```python
# Mudando o nome do atributo target CLASS_LABEL para label
df_initial_v1.rename(
    columns={"CLASS_LABEL": "label"},
    inplace=True
)
```

```python
print(df_initial_v1.shape)

print(df_initial_v1.columns.tolist())
```

```python
# Verificando se os atributos do df_inital_v1 são os mesmo do df_united_v1
set(df_initial_v1.columns) == set(df_built_v1.columns)
```

```python
df_initial_v1.to_csv(
    "../datasets/processed/df_initial_v1.csv",
    index=False
)

df_built_v1.to_csv(
    "../datasets/processed/df_united_v1.csv",
    index=False
)
```
