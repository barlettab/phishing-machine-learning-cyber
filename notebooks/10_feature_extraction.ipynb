```python
import sys
import os

sys.path.append(
    os.path.abspath("..")
)

from module.feature_extractor_v2 import extract_features
import pandas as pd
```

```python
df = pd.read_csv("../data/datasetURL.csv")
```

```python
df.head()
```

```python
# 5. Gerar features
features_df = pd.DataFrame(
    [extract_features(url) for url in df["url"]]
)
```

```python
features_df.head()
```

```python
features_df.shape
```

```python
features_df.info()
```

```python
features_df.isnull().sum()
```

```python
df_features = pd.concat(
    [features_df, df["label"]],
    axis=1
)
```

```python
df_features.head()
```

```python
df_features.to_csv(
    "../data/dataset_features.csv",
    index=False
)
```

```python
feature_stats = (
    df_features
    .groupby("label")
    .mean()
    .T
)

feature_stats
```

```python
corr = (
    df_features
    .drop(columns=["label"])
    .corr()
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.show()
```
