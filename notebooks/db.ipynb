```python
import pandas as pd

df_tranco = pd.read_csv(
    "../buildingdb/tranco.csv",
    header=None,
    names=["rank", "domain"]
)

df_tranco["url"] = "https://" + df_tranco["domain"]
df_tranco["label"] = 0
df_tranco["source"] = "tranco"

df_tranco = df_tranco[
    ["url", "label", "source"]
]
```

```python
df_tranco.head()
```

```python
with open(
    "../buildingdb/phishingGit.txt",
    "r",
    encoding="utf-8",
    errors="ignore"
) as f:

    urls = [
        linha.strip()
        for linha in f
        if linha.strip()
    ]

df_phishingdb = pd.DataFrame({
    "url": urls
})

df_phishingdb["label"] = 1
df_phishingdb["source"] = "phishingdb"
```

```python
df_phishingdb.shape
```

```python
print(len(df_phishingdb))
print(df_phishingdb["url"].nunique())
```

```python
df_phishingdb.sample(20, random_state=42)
```

```python
N = 57967
df_legit = df_tranco.sample(
    n=N,
    random_state=42
)

df_phish = df_phishingdb.sample(
    n=N,
    random_state=42
)
```

```python
df_final = pd.concat(
    [df_legit, df_phish],
    ignore_index=True
)

df_final = df_final.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```
