# Modelagem Preditiva para Detecção de Phishing

Neste notebook serão treinados e avaliados modelos de Machine Learning para classificação de URLs legítimas e maliciosas.

Inicialmente será utilizado o algoritmo Random Forest como modelo baseline, permitindo avaliar o desempenho do conjunto de features extraídas e identificar as características mais relevantes para a detecção de phishing.

#### Importação das bibliotecas

```python
# Importação das bibliotecas

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

#### Carregamento do dataset

```python
df = pd.read_csv("../data/dataset_features.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

#### Separação de Feature e Target

```python
X = df.drop(
    columns=[
        "label",
    ]
)

y = df["label"]
```

```python
X.shape
```

```python
y.value_counts()
```

#### Divisão Treino/Teste

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
y_train.value_counts(normalize=True)
```

```python
y_test.value_counts(normalize=True)
```

#### Random Forest

```python
rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

#### Predições

```python
y_pred = rf.predict(X_test)
```

#### Avaliações

```python
metrics_df = pd.DataFrame({
    "Métrica": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],
    "Valor": [
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred)
    ]
})

metrics_df
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(6,4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predito")
plt.ylabel("Real")
plt.title("Matriz de Confusão")
plt.show()
```

```python
importance = pd.Series(
    rf.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

importance.head(20)
```

```python
plt.figure(figsize=(10,8))

importance.head(20).sort_values().plot.barh()

plt.title(
    "Top 20 Features Mais Importantes"
)

plt.show()
```
